# Create one video

```python
df_categories = load_category_data()
df_examples = load_examples_data()
video_configs = load_video_configs()
```

```python
start_time = time.time()
cat_start_stop_idx = (41, 65)
df_categories_use = df_categories[
    (df_categories['id'] >= cat_start_stop_idx[0]) &
    (df_categories['id'] <= cat_start_stop_idx[1])
]
df_categories_use
```

```python
for i_catrow, catrow in df_categories_use.iterrows():
    chosen_category = catrow['chinese']
    print(f'!!!!!!! {i_catrow} START: {chosen_category}: {(time.time() - start_time):.2f} !!!!!!!')
    data_settings = load_data_settings(df_categories, chosen_category)
    create_directories(data_settings)
    df_filt = df_examples[df_examples['category'] == data_settings['category']].reset_index(drop=True)
    example_words = df_filt['chinese'].values
    # generate_and_zip_audio_files(audio_settings['voice_name_zh'], chosen_category, example_words)
    df_durations = stitch_audios(audio_settings, data_settings, example_words)
    no_hl_img_file_path = draw_vocab_list_whole_image(video_configs, df_categories, data_settings, df_filt)
    create_video_without_highlights(data_settings, video_configs, no_hl_img_file_path)
    create_video_with_highlights(df_durations, audio_settings, data_settings, video_configs)
```
